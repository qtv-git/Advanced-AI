#### c) Cho hai đỉnh A, B. Kiểm tra 2 đỉnh kề nhau?

```python
def DinhKe(G: dict, a: str, b: str):
    # Kiểm tra tồn tại
    if a not in G.keys() or b not in G.keys():
        return -1

    # print(G.get(a))
    # b nằm trong dsk của a
    for adj_1 in G.get(a):
        if b == adj_1:
            return 1
        
    # Nếu b không có trong dsk của a, kiểm tra a nằm trong dsk của b
    for adj_2 in G.get(b):
        if a == adj_2:
            return 1
        
    return 0
                    
print('B ke I: ' + str(DinhKe(ADJ, 'B', 'I')))
print('Z ke E: ' + str(DinhKe(ADJ, 'Z', 'E')))
print('H ke A: ' + str(DinhKe(ADJ, 'H', 'A')))
print('B ke A: ' + str(DinhKe(ADJ, 'B', 'A')))
print('K ke H: ' + str(DinhKe(ADJ, 'K', 'H')))
```

#### d) Cho một đỉnh a. Hỏi danh sách các đỉnh kề với a

```python
def LayDinhKe(G: dict, a: str):
    if a not in G.keys():
        return None
    
    return list(G.get(a))

Dinh_test = ['A', 'F', 'Z', 'G']
for dinh in Dinh_test:
    if LayDinhKe(ADJ, dinh) == None:
        print(f'Dinh {dinh} khong ton tai')
    elif LayDinhKe(ADJ, dinh) == []:
        print(f'{dinh} khong ke voi dinh nao')
    else: print(f'Dinh {dinh} ke voi: {LayDinhKe(ADJ, dinh)}')
```

```python
def LuuDoThi(G: dict, file_path, verbose = True):
    import os

    # Tạo thư mục chứa file_path
    file_dir = os.path.dirname(file_path)
    if file_dir != "" and os.path.exists(file_path) == False:
        os.makedirs(file_dir)
        if verbose: print(f"+ Tao thu muc: {file_dir}")
    
    # Lưu đồ thị
    with open(file_path, "wt") as file:
        file.write(str(len(G)) + '\n')

        # text = ' '.join(map(str, s)): chuyển các phần tử trong set thành chuỗi và phân cách bởi ' '
        for item in G.items():
            file.write(f'{item[0]} {' '.join(map(str, item[1]))}\n')

        if verbose: print(f"Luu do thi thanh cong xuong tap tin: {file_path}")

LuuDoThi(ADJ, 'dske/dothi.txt', verbose = True)
with open('dske/dothi.txt', "rt") as file:
    lines = file.readlines()
    for line in lines: print(line, end="")
```

#### 3.2 Đọc đồ thị

- read(): đọc toàn bộ file
- readline(): đọc từng dòng
- readlines(): đọc toàn bộ file, mỗi dòng là một phần tử của list

```python
import pprint

def DocDoThi(file_path, verbose):
    import os
    result = None
    if os.path.exists(file_path):
        G = {}
        with open(file_path, 'rt') as file:
            n = int(file.readline())
            for _ in range(n):
                # Sử dụng split() để loại bỏ khoảng trắn kể cả \n \t
                line = file.readline().split()
                # Lấy giá trị đầu là key, các giá trị còn lại là values
                G[line[0]] = set(line[1:])

        result = G
    else:
        result = None
    return result

G = DocDoThi('dske/dothi.txt', True)
pprint.pprint(G)
```

# B2. Tìm kiếm đường đi trên đồ thị

### 1. TÌm kiếm theo chiều rộng (BFS)

```python
import pprint

def BFS(G, start, goal):
    if start not in G or goal not in G:
        return None

    path = {start: None}      # path[v] = cha của v
    s_open = [start]          # hàng đợi (queue)
    s_closed = []             # các đỉnh đã thăm

    while s_open:
        u = s_open.pop(0)     # lấy đỉnh đầu hàng
        s_closed.append(u)

        for v in G[u]:
            if v not in path:
                path[v] = u
                s_open.append(v)

    return path

path_bfs = BFS(ADJ, "A", "B")
pprint.pprint(path_bfs)
```

```python
def find_path(path, start, goal):
    result = []
    u = goal

    while u is not None:
        if u != start:
            result.append(u)  # Thêm goal vao result
            u = path.get(u)  # Lấy đỉnh cha của u
            if u == None:
                result = []
                break
        else:
            result.append(u)
            break

    result = list(reversed(result))  # Lật ngược list từ start đến end

    return result
```

```python
result = find_path(path_bfs, 'A', 'K')

if result:
    print(result)
else:
    print('Khong tim thay duong')
```
